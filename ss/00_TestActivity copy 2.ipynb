#### 0. Import libraries

```python
import simpy
import shapely.geometry

import openclsim.core as core
import openclsim.model as model
```

#### 1. Initialise simpy environment

```python
# setup environment
my_env = simpy.Environment(initial_time=0)
registry = {}
```

#### 2. Define object classes

```python
#geometry
quay_loc = shapely.geometry.Point(4.18055556, 52.18664444)
vessel_loc = shapely.geometry.Point(4.18055556, 52.18664444)
yard_loc = shapely.geometry.Point(4.25222222, 52.11428333)
```

```python
#create a Site object based on desired mixin classes
Site = type (
    "Quay",
    (
        core.Identifiable,
        core.Log,
        core.Locatable,
        core.HasContainer,
        core.HasResource,
    ),
    {},
)

#create a TransportProcessingResource object based on desired mixin classes 
VesselUnloader = type(
    "QuayCrane",
    (
        core.Identifiable,
        core.ContainerDependentMovable,
        core.HasResource,
        core.Processor,
        #core.Log,
        core.Locatable,
        core.LoadingFunction,
        core.UnloadingFunction,
    ),
    {},
)

LandTransporter = type(
    "PrimeMover",
    (
        core.Identifiable,
        core.ContainerDependentMovable,
        core.Processor,
        core.HasResource,
        #core.Log,
        core.Locatable,
        core.LoadingFunction,
        core.UnloadingFunction,
    ),
    {},
)
```

##### 3.3 Create activity/activities

```python
#discharge until vessel is empty
qc01_discharge_step = model.ShiftAmountActivity(
    env=my_env,
    name="QC01_discharge_step",          # <-- important: used as trigger "name"
    registry=registry,
    processor=qc01,
    origin=vessel01,
    destination=quay_site,           # local co-location: destination must be at same geometry as origin
    amount=1,
    duration=qc_cycle_time,
    )
```

```python
def make_mover_cycle(mover, idx):

    pm_load = model.ShiftAmountActivity(
        env=my_env,
        name=f"PM{idx}_load",
        registry=registry,

        processor=mover,
        origin=quay_site,
        destination=mover,

        amount=1,
        duration=pm_load_time,
    )

    pm_travel_to_yard = model.MoveActivity(
        env=my_env,
        name=f"PM{idx}_to_yard",
        registry=registry,

        mover=mover,
        destination=yard_block,
    )

    pm_unload = model.ShiftAmountActivity(
        env=my_env,
        name=f"PM{idx}_unload",
        registry=registry,

        processor=mover,
        origin=mover,
        destination=yard_block,

        amount=1,
        duration=pm_unload_time,
    )

    pm_return = model.MoveActivity(
        env=my_env,
        name=f"PM{idx}_return",
        registry=registry,

        mover=mover,
        destination=quay_site,
    )

    cycle = model.SequentialActivity(
        env=my_env,
        name=f"PM{idx}_cycle",
        registry=registry,

        sub_processes=[
            pm_load,
            pm_travel_to_yard,
            pm_unload,
            pm_return,
        ],
    )

    return model.WhileActivity(
        env=my_env,
        name=f"PM{idx}_loop",
        registry=registry,

        condition_event=[
            {
                "type": "container",
                "concept": vessel01,
                "state": "gt",
                "value": 0,
            }
        ],

        sub_processes=[cycle],
    )
```

```python
mover_processes = [
    make_mover_cycle(pm, i + 1)
    for i, pm in enumerate(pm_fleet)
]

for pm in pm_fleet:
    print (f"{pm.name} initial location: {pm.geometry}")
```

```python
qc_discharge_loop = model.WhileActivity(
    env=my_env,
    name="QC_discharge_loop",
    registry=registry,
    sub_processes=[qc01_discharge_step],
    condition_event=[
        {"type": "container", "concept": vessel01, "state": "not_empty"}  # or your correct stop condition
    ],
)
```

```python
# Run QC loop + all mover cycles in parallel
system = model.ParallelActivity(
    env=my_env,
    name="terminal_fleet_system",
    registry=registry,
    sub_processes=[qc_discharge_loop, *mover_processes],
)
```

#### 4. Register processes and run simpy

```python
model.register_processes([system])
my_env.run()
```
