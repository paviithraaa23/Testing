```python
qc_cycle_time = 240
pm_load_time = 120
pm_unload_time = 120

pm_speed = 5
fleet_size = 10
```

```python
def constant_speed(v):
    return lambda x: v
```

```python
def create_pm_fleet(
    env,
    registry,
    size,
    geometry,
    speed=5,
    capacity=1,
):

    fleet = []

    for i in range(size):

        pm = LandTransporter(
            env=env,
            name=f"pm_{i+1}",
            geometry=geometry,

            capacity=capacity,

            compute_v=lambda x, v=speed: v,

            loading_rate=1,
            unloading_rate=1,

            nr_resources=1,
        )

        fleet.append(pm)

    return fleet
```

```python
# initialise registry
registry = {}

pm_fleet = create_pm_fleet(
    env=my_env,
    registry=registry,
    size=fleet_size,
    geometry=quay_loc,
)
```

##### 3.3 Create activity/activities

```python
#discharge until vessel is empty
qc_discharge_step = model.ShiftAmountActivity(
    env=my_env,
    name="QC_discharge_step",          # <-- important: used as trigger "name"
    registry=registry,
    processor=qc,
    origin=vessel,
    destination=quay_site,           # local co-location: destination must be at same geometry as origin
    amount=1,
    duration=qc_cycle_time,
    )
```

```python
def make_mover_process(mover, mover_index: int):
    lm_load = model.ShiftAmountActivity(
        env=my_env,
        name=f"LM{mover_index}_load_step",
        registry=registry,
        processor=mover,
        origin=quay_site,
        destination=mover,          # local co-location required
        amount=1,
        duration=pm_load_time,
        id_="TEU_1",
        start_event=[
            {"name": "QC_discharge_step", "type": "activity", "state": "done"}
        ],
    )

    lm_travel = model.MoveActivity(
        env=my_env,
        name=f"LM{mover_index}_travel_to_yard",
        registry=registry,
        mover=mover,
        destination=yard_block,
    )

    lm_unload = model.ShiftAmountActivity(
        env=my_env,
        name=f"LM{mover_index}_unload_step",
        registry=registry,
        processor=mover,
        origin=mover,
        destination=yard_block,     # local co-location required
        amount=1,
        duration=pm_unload_time,
        id_="TEU_1",
    )

    lm_cycle = model.SequentialActivity(
        env=my_env,
        name=f"LM{mover_index}_cycle",
        registry=registry,
        sub_processes=[lm_load, lm_travel, lm_unload],
    )

    return lm_cycle
```

```python
def make_mover_cycle(mover, idx):

    lm_load = model.ShiftAmountActivity(
        env=env,
        name=f"LM{idx}_load",
        registry=registry,

        processor=mover,
        origin=quay_site,
        destination=mover,

        amount=1,
        duration=lm_load_duration,
    )

    lm_travel_to_yard = model.MoveActivity(
        env=env,
        name=f"LM{idx}_to_yard",
        registry=registry,

        mover=mover,
        destination=yard_site,
    )

    lm_unload = model.ShiftAmountActivity(
        env=env,
        name=f"LM{idx}_unload",
        registry=registry,

        processor=mover,
        origin=mover,
        destination=yard_site,

        amount=1,
        duration=lm_unload_duration,
    )

    lm_return = model.MoveActivity(
        env=env,
        name=f"LM{idx}_return",
        registry=registry,

        mover=mover,
        destination=quay_site,
    )

    cycle = model.SequentialActivity(
        env=env,
        name=f"LM{idx}_cycle",
        registry=registry,

        sub_processes=[
            lm_load,
            lm_travel_to_yard,
            lm_unload,
            lm_return,
        ],
    )

    return model.WhileActivity(
        env=env,
        name=f"LM{idx}_loop",
        registry=registry,

        condition_event=[
            {
                "type": "container",
                "concept": vessel01,
                "state": "gt",
                "value": 0,
            }
        ],

        sub_processes=[cycle],
    )
```

```python
mover_processes = [
    make_mover_cycle(pm, i + 1)
    for i, pm in enumerate(pm_fleet)
]

for pm in pm_fleet:
    print (f"{pm.name} initial location: {pm.geometry}")
```

```python
qc_discharge_loop = model.WhileActivity(
    env=my_env,
    name="QC_discharge_loop",
    registry=registry,
    sub_processes=[qc_discharge_step],
    condition_event=[
        {"type": "container", "concept": vessel, "state": "not_empty"}  # or your correct stop condition
    ],
)
```

```python
# Run QC loop + all mover cycles in parallel
system = model.ParallelActivity(
    env=my_env,
    name="terminal_fleet_system",
    registry=registry,
    sub_processes=[qc_discharge_loop, *mover_processes],
)
```

#### 4. Register processes and run simpy

```python
model.register_processes([system])
my_env.run()
```
